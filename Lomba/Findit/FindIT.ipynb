```python
# install dan import semua library
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import timm
from PIL import Image
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score
import pandas as pd
import numpy as np
import os
from collections import Counter

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using: {device}')
```

```python
# siapkan dataset
# Buat dataframe dari folder train
train_path = './dataset/train'
data = []
for label in os.listdir(train_path):
    folder = os.path.join(train_path, label)
    for img_name in os.listdir(folder):
        data.append({'path': os.path.join(folder, img_name), 'label': label})

df = pd.DataFrame(data)
print(df['label'].value_counts())

# Label encoding
label2idx = {l: i for i, l in enumerate(sorted(df['label'].unique()))}
idx2label = {i: l for l, i in label2idx.items()}
df['label_idx'] = df['label'].map(label2idx)

print('\nLabel mapping:')
for k, v in label2idx.items():
    print(f'  {k}: {v}')
```

```python
# dataset class dan augmentasi
class FaceDataset(Dataset):
    def __init__(self, df, transform=None):
        self.df = df.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = Image.open(row['path']).convert('RGB')
        if self.transform:
            img = self.transform(img)
        return img, row['label_idx']

# Augmentasi untuk training
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1),
    transforms.RandomAffine(degrees=0, translate=(0.1, 0.1)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
    transforms.RandomErasing(p=0.3),
])

# Tanpa augmentasi untuk validasi
val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])
```

```python
## Training function ##
def train_one_fold(df, train_idx, val_idx, fold_num):
    print(f'\n===== FOLD {fold_num} =====')

    train_df = df.iloc[train_idx]
    val_df = df.iloc[val_idx]

    train_dataset = FaceDataset(train_df, train_transform)
    val_dataset = FaceDataset(val_df, val_transform)

    # Weighted sampler untuk handle class imbalance
    class_counts = Counter(train_df['label_idx'].values)
    weights = [1.0 / class_counts[label] for label in train_df['label_idx'].values]
    sampler = torch.utils.data.WeightedRandomSampler(weights, len(weights))

    train_loader = DataLoader(train_dataset, batch_size=32, sampler=sampler, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=2)

    # Model — EfficientNet B0
    model = timm.create_model('efficientnet_b0', pretrained=True, num_classes=6)
    model = model.to(device)

    # Class weights untuk loss
    total = sum(class_counts.values())
    class_weights = torch.tensor([total / (6 * class_counts[i]) for i in range(6)], dtype=torch.float).to(device)
    criterion = nn.CrossEntropyLoss(weight=class_weights)

    optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-2)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=15)

    best_f1 = 0
    best_model_state = None

    for epoch in range(15):
        # Train
        model.train()
        train_loss = 0
        for imgs, labels in train_loader:
            imgs, labels = imgs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(imgs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()
        scheduler.step()

        # Validate
        model.eval()
        all_preds, all_labels = [], []
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs = imgs.to(device)
                outputs = model(imgs)
                preds = outputs.argmax(dim=1).cpu().numpy()
                all_preds.extend(preds)
                all_labels.extend(labels.numpy())

        macro_f1 = f1_score(all_labels, all_preds, average='macro')
        print(f'Epoch {epoch+1:2d} | Loss: {train_loss/len(train_loader):.4f} | Macro F1: {macro_f1:.4f}')

        if macro_f1 > best_f1:
            best_f1 = macro_f1
            best_model_state = model.state_dict().copy()

    print(f'Best Fold {fold_num} Macro F1: {best_f1:.4f}')

    # Load best model
    model.load_state_dict(best_model_state)
    return model, best_f1
```

<h2> EDA 1 — Kenalan dengan Data

```python
# Pertama-tama, kita perlu tau dulu isi dataset kita
import pandas as pd
import os
from PIL import Image
from pathlib import Path

train_path = './dataset/train'
test_path = './dataset/test'

# Kumpulkan info setiap gambar
records = []
for cls in sorted(os.listdir(train_path)):
    folder = os.path.join(train_path, cls)
    for fname in os.listdir(folder):
        fpath = os.path.join(folder, fname)
        fsize = os.path.getsize(fpath) / 1024  # dalam KB
        ext = Path(fname).suffix.lower()
        try:
            img = Image.open(fpath)
            w, h = img.size
            mode = img.mode
        except:
            w, h, mode = None, None, 'corrupt'
        records.append({
            'filename': fname, 'class': cls, 'width': w, 'height': h,
            'aspect_ratio': round(w/h, 2) if h else None,
            'filesize_kb': round(fsize, 1), 'format': ext, 'color_mode': mode
        })

df_eda = pd.DataFrame(records)
print(f"Total gambar training: {len(df_eda)}")
print(f"Total gambar test: {len(os.listdir(test_path))}")
print(f"\nKolom yang tersedia:\n{list(df_eda.columns)}")
for cls in sorted(df_eda['class'].unique()):
    print(f"\n{'='*60}")
    print(f"  {cls}")
    print(f"{'='*60}")
    sub = df_eda[df_eda['class'] == cls]
    print(f"Jumlah gambar: {len(sub)}")
    print(f"Format: {sub['format'].value_counts().to_dict()}")
    print(f"Color mode: {sub['color_mode'].value_counts().to_dict()}")
    print(sub[['width', 'height', 'aspect_ratio', 'filesize_kb']].describe().round(1))
```

<h2> EDA 6 — Deteksi Duplikat

```python
# Gambar duplikat bisa bikin model bingung, apalagi kalau ada di kelas berbeda
import hashlib

def file_hash(filepath):
    with open(filepath, 'rb') as f:
        return hashlib.md5(f.read()).hexdigest()

hashes = {}
duplicates = []

for _, row in df_eda.iterrows():
    fpath = os.path.join(train_path, row['class'], row['filename'])
    h = file_hash(fpath)
    if h in hashes:
        duplicates.append({
            'file_1': hashes[h]['filename'],
            'class_1': hashes[h]['class'],
            'file_2': row['filename'],
            'class_2': row['class'],
            'same_class': hashes[h]['class'] == row['class']
        })
    else:
        hashes[h] = row

print(f"Total duplikat ditemukan: {len(duplicates)}")
if len(duplicates) > 0:
    dup_df = pd.DataFrame(duplicates)
    cross_class = dup_df[dup_df['same_class'] == False]
    same_class = dup_df[dup_df['same_class'] == True]
    print(f"  - Duplikat dalam kelas yang sama: {len(same_class)}")
    print(f"  - Duplikat BEDA kelas (bahaya!): {len(cross_class)}")
    if len(cross_class) > 0:
        print("\n⚠ Duplikat beda kelas:")
        print(cross_class)
else:
    print("Tidak ada duplikat — bagus!")
```

```python
# Breakdown duplikat beda kelas per pasangan kelas
dup_df = pd.DataFrame(duplicates)
cross = dup_df[dup_df['same_class'] == False]

# Hitung per pasangan kelas
pair_counts = {}
for _, row in cross.iterrows():
    pair = tuple(sorted([row['class_1'], row['class_2']]))
    pair_counts[pair] = pair_counts.get(pair, 0) + 1

print("Duplikat beda kelas per pasangan:")
for pair, count in sorted(pair_counts.items(), key=lambda x: -x[1]):
    print(f"  {pair[0]} ↔ {pair[1]}: {count} gambar")
```

```python
### super train restart ###
# === SEMUA DALAM SATU CELL ===
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import timm
from PIL import Image
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score, confusion_matrix, classification_report
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
from collections import Counter

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using: {device}')

# === Dataset ===
train_path = './dataset/train'
data = []
for label in os.listdir(train_path):
    folder = os.path.join(train_path, label)
    for img_name in os.listdir(folder):
        data.append({'path': os.path.join(folder, img_name), 'label': label})

df = pd.DataFrame(data)
label2idx = {l: i for i, l in enumerate(sorted(df['label'].unique()))}
idx2label = {i: l for l, i in label2idx.items()}
df['label_idx'] = df['label'].map(label2idx)
print(df['label'].value_counts())

class FaceDataset(Dataset):
    def __init__(self, df, transform=None):
        self.df = df.reset_index(drop=True)
        self.transform = transform
    def __len__(self):
        return len(self.df)
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = Image.open(row['path']).convert('RGB')
        if self.transform:
            img = self.transform(img)
        return img, row['label_idx']

train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1),
    transforms.RandomAffine(degrees=0, translate=(0.1, 0.1)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
    transforms.RandomErasing(p=0.3),
])

val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])

# === Training Function ===
def train_one_fold(df, train_idx, val_idx, fold_num):
    print(f'\n===== FOLD {fold_num} =====')
    train_df = df.iloc[train_idx]
    val_df = df.iloc[val_idx]

    train_dataset = FaceDataset(train_df, train_transform)
    val_dataset = FaceDataset(val_df, val_transform)

    class_counts = Counter(train_df['label_idx'].values)
    weights = [1.0 / class_counts[label] for label in train_df['label_idx'].values]
    sampler = torch.utils.data.WeightedRandomSampler(weights, len(weights))

    train_loader = DataLoader(train_dataset, batch_size=32, sampler=sampler, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=2)

    model = timm.create_model('efficientnet_b0', pretrained=True, num_classes=6)
    model = model.to(device)

    total = sum(class_counts.values())
    class_weights = torch.tensor([total / (6 * class_counts[i]) for i in range(6)], dtype=torch.float).to(device)
    criterion = nn.CrossEntropyLoss(weight=class_weights)
    optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-2)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=15)

    best_f1 = 0
    best_model_state = None

    for epoch in range(15):
        model.train()
        train_loss = 0
        for imgs, labels in train_loader:
            imgs, labels = imgs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(imgs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()
        scheduler.step()

        model.eval()
        all_preds, all_labels = [], []
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs = imgs.to(device)
                outputs = model(imgs)
                preds = outputs.argmax(dim=1).cpu().numpy()
                all_preds.extend(preds)
                all_labels.extend(labels.numpy())

        macro_f1 = f1_score(all_labels, all_preds, average='macro')
        print(f'Epoch {epoch+1:2d} | Loss: {train_loss/len(train_loader):.4f} | Macro F1: {macro_f1:.4f}')

        if macro_f1 > best_f1:
            best_f1 = macro_f1
            best_model_state = model.state_dict().copy()

    print(f'Best Fold {fold_num} Macro F1: {best_f1:.4f}')
    model.load_state_dict(best_model_state)
    return model, best_f1

# === 5-Fold Training ===
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
models = []
scores = []

for fold, (train_idx, val_idx) in enumerate(skf.split(df, df['label_idx'])):
    model, score = train_one_fold(df, train_idx, val_idx, fold + 1)
    models.append(model)
    scores.append(score)

print(f'\n===== HASIL =====')
print(f'Average Macro F1: {np.mean(scores):.4f} (+/- {np.std(scores):.4f})')

# === Confusion Matrix ===
folds = list(skf.split(df, df['label_idx']))
_, val_idx = folds[-1]
val_df = df.iloc[val_idx]
val_dataset = FaceDataset(val_df, val_transform)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)

all_probs = []
for model in models:
    model.eval()
    fold_probs = []
    with torch.no_grad():
        for imgs, _ in val_loader:
            imgs = imgs.to(device)
            outputs = torch.softmax(model(imgs), dim=1)
            fold_probs.append(outputs.cpu().numpy())
    all_probs.append(np.concatenate(fold_probs))

avg_probs = np.mean(all_probs, axis=0)
preds = np.argmax(avg_probs, axis=1)
true_labels = val_df['label_idx'].values

print(classification_report(true_labels, preds, target_names=[idx2label[i] for i in range(6)]))

cm = confusion_matrix(true_labels, preds)
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=[idx2label[i] for i in range(6)],
            yticklabels=[idx2label[i] for i in range(6)])
plt.title('Confusion Matrix (Ensemble)')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.tight_layout()
plt.show()
```

```python
### improved training ###
# === IMPROVED VERSION ===
# === VERSI 3: Kembali ke setup yang works + targeted improvement ===
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import timm
from PIL import Image
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score, confusion_matrix, classification_report
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
from collections import Counter

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using: {device}')

train_path = './dataset/train'
data = []
for label in os.listdir(train_path):
    folder = os.path.join(train_path, label)
    for img_name in os.listdir(folder):
        data.append({'path': os.path.join(folder, img_name), 'label': label})

df = pd.DataFrame(data)
label2idx = {l: i for i, l in enumerate(sorted(df['label'].unique()))}
idx2label = {i: l for l, i in label2idx.items()}
df['label_idx'] = df['label'].map(label2idx)

class FaceDataset(Dataset):
    def __init__(self, df, transform=None):
        self.df = df.reset_index(drop=True)
        self.transform = transform
    def __len__(self):
        return len(self.df)
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = Image.open(row['path']).convert('RGB')
        if self.transform:
            img = self.transform(img)
        return img, row['label_idx']

# Augmentasi moderat (seperti V1 tapi sedikit lebih kuat)
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.05),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
    transforms.RandomErasing(p=0.2),
])

val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
])

def train_one_fold(df, train_idx, val_idx, fold_num):
    print(f'\n===== FOLD {fold_num} =====')
    train_df = df.iloc[train_idx]
    val_df = df.iloc[val_idx]

    train_dataset = FaceDataset(train_df, train_transform)
    val_dataset = FaceDataset(val_df, val_transform)

    class_counts = Counter(train_df['label_idx'].values)
    weights = [1.0 / class_counts[label] for label in train_df['label_idx'].values]
    sampler = torch.utils.data.WeightedRandomSampler(weights, len(weights))

    train_loader = DataLoader(train_dataset, batch_size=32, sampler=sampler, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=2)

    # EfficientNet B0 (yang terbukti lebih bagus)
    model = timm.create_model('efficientnet_b0', pretrained=True, num_classes=6)
    model = model.to(device)

    total = sum(class_counts.values())
    class_weights = torch.tensor([total / (6 * class_counts[i]) for i in range(6)], dtype=torch.float).to(device)
    criterion = nn.CrossEntropyLoss(weight=class_weights, label_smoothing=0.05)

    # 2-stage LR: backbone kecil, head besar
    backbone_params = [p for n, p in model.named_parameters() if 'classifier' not in n]
    head_params = [p for n, p in model.named_parameters() if 'classifier' in n]
    optimizer = torch.optim.AdamW([
        {'params': backbone_params, 'lr': 3e-5},
        {'params': head_params, 'lr': 1e-3},
    ], weight_decay=1e-2)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=20)

    best_f1 = 0
    best_model_state = None
    patience = 0

    for epoch in range(20):
        model.train()
        train_loss = 0
        for imgs, labels in train_loader:
            imgs, labels = imgs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(imgs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()
        scheduler.step()

        model.eval()
        all_preds, all_labels = [], []
        with torch.no_grad():
            for imgs, labels in val_loader:
                imgs = imgs.to(device)
                outputs = model(imgs)
                preds = outputs.argmax(dim=1).cpu().numpy()
                all_preds.extend(preds)
                all_labels.extend(labels.numpy())

        macro_f1 = f1_score(all_labels, all_preds, average='macro')
        print(f'Epoch {epoch+1:2d} | Loss: {train_loss/len(train_loader):.4f} | Macro F1: {macro_f1:.4f}')

        if macro_f1 > best_f1:
            best_f1 = macro_f1
            best_model_state = model.state_dict().copy()
            patience = 0
        else:
            patience += 1
            if patience >= 7:
                print(f'Early stopping at epoch {epoch+1}')
                break

    print(f'Best Fold {fold_num} Macro F1: {best_f1:.4f}')
    model.load_state_dict(best_model_state)
    return model, best_f1

# === Training ===
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
models = []
scores = []

for fold, (train_idx, val_idx) in enumerate(skf.split(df, df['label_idx'])):
    model, score = train_one_fold(df, train_idx, val_idx, fold + 1)
    models.append(model)
    scores.append(score)

print(f'\n===== HASIL V3 =====')
print(f'Average Macro F1: {np.mean(scores):.4f} (+/- {np.std(scores):.4f})')

# === Confusion Matrix ===
folds = list(skf.split(df, df['label_idx']))
_, val_idx = folds[-1]
val_df = df.iloc[val_idx]
val_dataset = FaceDataset(val_df, val_transform)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)

all_probs = []
for model in models:
    model.eval()
    fold_probs = []
    with torch.no_grad():
        for imgs, _ in val_loader:
            imgs = imgs.to(device)
            outputs = torch.softmax(model(imgs), dim=1)
            fold_probs.append(outputs.cpu().numpy())
    all_probs.append(np.concatenate(fold_probs))

avg_probs = np.mean(all_probs, axis=0)
preds = np.argmax(avg_probs, axis=1)
true_labels = val_df['label_idx'].values

print(classification_report(true_labels, preds, target_names=[idx2label[i] for i in range(6)]))

cm = confusion_matrix(true_labels, preds)
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=[idx2label[i] for i in range(6)],
            yticklabels=[idx2label[i] for i in range(6)])
plt.title('Confusion Matrix - V3')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.tight_layout()
plt.show()
```
