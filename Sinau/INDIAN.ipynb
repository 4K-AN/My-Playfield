Pada tahap awal, kita membuat code seperti di bawah ini. Hal ini berfungsi untuk memuat dataset Indian Liver Patient dari repository GitHub dan melakukan setup awal untuk analisis data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# URL dari file CSV di GitHub (versi 'raw')
url = 'https://raw.githubusercontent.com/4K-AN/My-Playfield/main/Tugas/DATSI/DATASET/indian_liver_patient.csv'

# Membaca dataset menggunakan pandas
try:
    df = pd.read_csv(url)
    print("✅ Dataset berhasil dimuat.")
except Exception as e:
    print(f"❌ Gagal memuat dataset. Error: {e}")

# Menampilkan 5 baris pertama untuk memastikan data termuat dengan benar
print("\nLima baris pertama dari dataset:")
df.head()
```

```python

print("Lima baris pertama dataset:")
display(df.head())

print("\nInformasi Dataset:")
df.info()

print("\nStatistik Deskriptif:")
display(df.describe())

print("\nJumlah Nilai yang Hilang per Kolom:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8, 6))
ax = sns.countplot(x='Dataset', data=df, palette=['#35B779', '#FDE725'])
plt.title('Distribusi Pasien Penyakit Hati', fontsize=15)
plt.xlabel('Status Penyakit Hati', fontsize=12)
plt.ylabel('Jumlah Pasien', fontsize=12)
plt.xticks(ticks=[0, 1], labels=['Tidak Sakit', 'Sakit Hati'])

# Menambahkan label angka di atas bar
for p in ax.patches:
    ax.annotate(f'{p.get_height()}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 9), textcoords='offset points')

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Distribusi Penyakit Hati berdasarkan Jenis Kelamin
sns.countplot(x='Gender', hue='Dataset', data=df, ax=axes[0], palette=['#440154', '#21908C'])
axes[0].set_title('Penyakit Hati Berdasarkan Jenis Kelamin')
axes[0].set_xlabel('Jenis Kelamin')
axes[0].set_ylabel('Jumlah Pasien')
axes[0].set_xticks([0, 1], ['Perempuan', 'Laki-laki'])

# Plot 2: Distribusi Usia
sns.histplot(data=df, x='Age', hue='Dataset', kde=True, multiple='stack', ax=axes[1], palette=['#440154', '#21908C'])
axes[1].set_title('Distribusi Usia Pasien')
axes[1].set_xlabel('Usia (Tahun)')
axes[1].set_ylabel('Frekuensi')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(), annot=True, cmap='viridis', fmt='.2f', linewidths=0.5)
plt.title('Matriks Korelasi Fitur-Fitur Biokimia', fontsize=16)
plt.show()
```
