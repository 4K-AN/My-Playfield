## 2. Memahami Faktorisasi

**Faktor** adalah bilangan bulat yang membagi habis bilangan lain tanpa sisa.

Contoh untuk angka 10:
- 10 ÷ 1 = 10 (habis dibagi, jadi 1 dan 10 adalah faktor)
- 10 ÷ 2 = 5 (habis dibagi, jadi 2 dan 5 adalah faktor)
- 10 ÷ 3 = 3.33... (tidak habis, jadi 3 bukan faktor)
- dst...

**Faktor dari 10 adalah: 1, 2, 5, 10**

## 3. Menemukan Semua Faktor dari Sebuah Bilangan

Mari kita buat fungsi untuk menemukan semua faktor dari bilangan yang diberikan.

```python
def cari_faktor(n):
    """
    Mencari semua faktor dari bilangan n
    
    Parameters:
    n (int): Bilangan yang akan dicari faktornya
    
    Returns:
    list: Daftar semua faktor dari n
    """
    faktor = []
    for i in range(1, n + 1):
        if n % i == 0:  # Jika sisa pembagian adalah 0
            faktor.append(i)
    return faktor

# Contoh: Cari faktor dari 10
bilangan = 10
hasilFaktor = cari_faktor(bilangan)
print(f"Faktor dari {bilangan}: {hasilFaktor}")
print(f"Jumlah faktor: {len(hasilFaktor)}")
```

```python
def cari_faktor_efisien(n):
    """
    Mencari faktor dengan cara yang lebih efisien
    Cukup cek sampai akar dari n
    """
    faktor = []
    for i in range(1, int(n**0.5) + 1):
        if n % i == 0:
            faktor.append(i)
            if i != n // i:  # Hindari duplikat pada bilangan kuadrat sempurna
                faktor.append(n // i)
    return sorted(faktor)

# Bandingkan kedua cara
print("\n=== Perbandingan Cara Mencari Faktor ===")
print(f"Cara 1 (iterasi semua): {cari_faktor(10)}")
print(f"Cara 2 (efisien): {cari_faktor_efisien(10)}")

# Test dengan bilangan yang lebih besar
print(f"\nFaktor dari 36: {cari_faktor_efisien(36)}")
print(f"Faktor dari 100: {cari_faktor_efisien(100)}")
```

## 4. Faktorisasi Prima

**Faktor Prima** adalah faktor yang merupakan bilangan prima (hanya bisa dibagi 1 dan dirinya sendiri).

Contoh:
- 10 = 2 × 5
- 36 = 2 × 2 × 3 × 3 = 2² × 3²
- 100 = 2 × 2 × 5 × 5 = 2² × 5²

```python
def faktorisasi_prima(n):
    """
    Menguraikan bilangan ke dalam faktor-faktor prima
    """
    faktor_prima = []
    pembagi = 2
    
    # Proses sampai n menjadi 1
    while n > 1:
        while n % pembagi == 0:
            faktor_prima.append(pembagi)
            n = n // pembagi
        pembagi += 1
        
        # Optimasi: jika pembagi > akar(n), maka n adalah prima
        if pembagi * pembagi > n and n > 1:
            faktor_prima.append(n)
            break
    
    return faktor_prima

# Contoh faktorisasi prima
print("=== Faktorisasi Prima ===")
contoh = [10, 36, 100, 24, 60]

for num in contoh:
    faktor = faktorisasi_prima(num)
    hasil_perkalian = " × ".join(map(str, faktor))
    print(f"{num} = {hasil_perkalian}")
```

## 5. Menyaring Faktor dengan Kondisi Logis

Kita dapat menggunakan kondisi logis untuk menyaring faktor yang memenuhi kriteria tertentu.

**Pertanyaan:** Dari angka 10, cari semua bilangan dari 2 hingga 9 yang membagi 10 habis.

Jawaban: Faktor dari 10 dalam rentang 2-9 adalah **2 dan 5**

```python
def saring_faktor(n, min_kondisi=1, max_kondisi=None):
    """
    Menyaring faktor dengan kondisi logis
    Mengembalikan faktor yang berada dalam rentang min_kondisi sampai max_kondisi
    """
    if max_kondisi is None:
        max_kondisi = n - 1
    
    faktor = []
    for i in range(min_kondisi, max_kondisi + 1):
        if n % i == 0:
            faktor.append(i)
    return faktor

# CONTOH KASUS: Dari angka 10, cari bilangan yang membagi habis
print("=== Contoh Kasus: Penggunaan Pertanyaan User ===")
angka = 10
print(f"\nAngka: {angka}")
print(f"Semua faktor dari {angka}: {cari_faktor_efisien(angka)}")

# Saring faktor dalam rentang 2 hingga 9
faktor_saring = saring_faktor(angka, min_kondisi=2, max_kondisi=9)
print(f"Faktor dari {angka} dalam rentang 2-9: {faktor_saring}")

# Contoh lain dengan berbagai kondisi
print("\n=== Berbagai Kondisi Penyaringan ===")
print(f"Faktor > 1: {saring_faktor(10, min_kondisi=2)}")
print(f"Faktor < 10: {saring_faktor(10, max_kondisi=9)}")
print(f"Faktor ganjil (1,3,5,7,9): {[i for i in saring_faktor(10, max_kondisi=10) if i % 2 == 1]}")
print(f"Faktor genap (2,4,6,8): {[i for i in cari_faktor_efisien(10) if i % 2 == 0]}")
```

## 6. Analisis Faktor Lanjutan

### Faktor Persekutuan Terbesar (GCD - Greatest Common Divisor)
Faktor terbesar yang dimiliki oleh dua atau lebih bilangan.

### Kelipatan Persekutuan Terkecil (LCM - Least Common Multiple)
Kelipatan terkecil yang dimiliki oleh dua atau lebih bilangan.

### Analisis Hubungan Faktor
Mempelajari hubungan antara faktor-faktor dan bilangan aslinya.

```python
# GCD dan LCM menggunakan math module
from math import gcd

def hitung_gcd(a, b):
    """Menghitung GCD dengan definisi matematika"""
    return gcd(a, b)

def hitung_lcm(a, b):
    """Menghitung LCM menggunakan hubungan: LCM(a,b) = (a*b) / GCD(a,b)"""
    return abs(a * b) // gcd(a, b)

def faktor_persekutuan(a, b):
    """Mencari faktor-faktor yang sama dari dua bilangan"""
    faktor_a = cari_faktor_efisien(a)
    faktor_b = cari_faktor_efisien(b)
    return sorted(list(set(faktor_a) & set(faktor_b)))

print("=== GCD dan LCM ===")
a, b = 10, 15
print(f"GCD ({a}, {b}) = {hitung_gcd(a, b)}")
print(f"LCM ({a}, {b}) = {hitung_lcm(a, b)}")

print(f"\n=== Faktor Persekutuan ===")
print(f"Faktor dari {a}: {cari_faktor_efisien(a)}")
print(f"Faktor dari {b}: {cari_faktor_efisien(b)}")
print(f"Faktor Persekutuan: {faktor_persekutuan(a, b)}")

print("\n=== Analisis Hubungan ===")
# Verifikasi: GCD harus menjadi faktor persekutuan terbesar
print(f"Faktor Persekutuan Terbesar = {faktor_persekutuan(a, b)[-1]} (sama dengan GCD)")

# Hubungan: GCD × LCM = a × b
print(f"\nVerifikasi rumus: GCD × LCM = a × b")
print(f"{hitung_gcd(a, b)} × {hitung_lcm(a, b)} = {hitung_gcd(a, b) * hitung_lcm(a, b)}")
print(f"{a} × {b} = {a * b}")
```

## 8. Solusi Latihan

```python
def adalah_prima(n):
    """Mengecek apakah n adalah bilangan prima"""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# SOAL 1: Faktor dari 24 dan identifikasi bilangan prima
print("=" * 50)
print("SOLUSI SOAL 1")
print("=" * 50)
faktor_24 = cari_faktor_efisien(24)
prima_24 = [f for f in faktor_24 if adalah_prima(f)]
print(f"Faktor dari 24: {faktor_24}")
print(f"Yang merupakan bilangan prima: {prima_24}")

# SOAL 2: Faktorisasi prima, GCD dan LCM dari 60 dan 90
print("\n" + "=" * 50)
print("SOLUSI SOAL 2")
print("=" * 50)
print(f"Faktorisasi prima 60: {faktorisasi_prima(60)}")
print(f"Faktorisasi prima 90: {faktorisasi_prima(90)}")
print(f"GCD(60, 90) = {hitung_gcd(60, 90)}")
print(f"LCM(60, 90) = {hitung_lcm(60, 90)}")

# SOAL 3: Bilangan antara 3-10 yang membagi habis 20
print("\n" + "=" * 50)
print("SOLUSI SOAL 3")
print("=" * 50)
hasil_soal3 = saring_faktor(20, min_kondisi=3, max_kondisi=10)
print(f"Bilangan 3-10 yang membagi habis 20: {hasil_soal3}")

# SOAL 4: Bilangan sempurna
print("\n" + "=" * 50)
print("SOLUSI SOAL 4")
print("=" * 50)

def adalah_bilangan_sempurna(n):
    """Mengecek apakah n adalah bilangan sempurna"""
    faktor = cari_faktor_efisien(n)
    jumlah_faktor = sum(faktor[:-1])  # Semua faktor kecuali dirinya sendiri
    return jumlah_faktor == n

def cari_bilangan_sempurna(batas):
    """Mencari semua bilangan sempurna hingga batas tertentu"""
    sempurna = []
    for i in range(2, batas + 1):
        if adalah_bilangan_sempurna(i):
            faktor = cari_faktor_efisien(i)
            print(f"{i} adalah bilangan sempurna: {' + '.join(map(str, faktor[:-1]))} = {i}")
            sempurna.append(i)
    return sempurna

print("Mencari bilangan sempurna hingga 30:")
bilangan_sempurna = cari_bilangan_sempurna(30)
print(f"\nBilangan sempurna yang ditemukan: {bilangan_sempurna}")
```
